#Dataset Information

Source: Kaggle - https://www.kaggle.com/datasets/preritmasih/delhi-metro-cleaned-dataset

Year / Timeline: Data collected during 2022 to 2024

Dataset Description:
TripID: Unique identifier for each metro trip

Date: Date on which the trip occurred

From_Station: Origin station of the trip

To_Station: Destination station of the trip

Distance_km: Distance covered during the trip (in kilometers)

Fare: Ticket fare charged per passenger.

Cost_per_passenger: Operational cost incurred per passenger.

Passengers: Total number of passengers for the trip.

Ticket_Type: Type of ticket used.

Remarks: Additional notes or trip-related observations.

Profit_per_passenger: Net profit earned per passenger.

Profit: Total profit generated from the trip.

## Stage 3 – EDA and Visualizations
The objective of this stage is to analyze travel patterns, revenue distribution, and profitability trends within the Delhi Metro dataset using univariate, bivariate, and multivariate analysis techniques. Visualizations are used to uncover passenger behavior, route performance, and financial insights, with each chart interpreted from a business and operational perspective.

```python
#Chart 1: Distribution of Ticket Type

#Chart Type: Countplot

plt.figure(figsize=(8, 6))
sns.countplot(x='Ticket_Type',hue='Ticket_Type', data=df,palette='Set2')
plt.title('Distribution of Ticket Types')
plt.xlabel('Ticket Types')
plt.ylabel('count of Passengers')
plt.show()
```

# Interpretation – Ticket Type Countplot

The chart displays the frequency of trips by ticket type.

Token usage is higher compared to Smart Card usage.

This indicates that a larger number of passengers prefer on Tokens for travel.

```python
#Chart 2: Profit Distribution Boxplot

#Chart Type: Boxplot
plt.figure(figsize=(8, 6))
sns.boxplot(y=df['Profit'])
plt.title('Profit Distribution')
plt.ylabel('Total Profit')
plt.grid(True)
plt.show()
```

# Interpretation – Profit Distribution Boxplot

The median profit is positive suggesting that most trips are profitable overall.

Negative outliers indicate a small number of trips with significant losses.

Positive outliers represent highly profitable trips, possibly due to higher passenger counts or longer distances.

The presence of outliers on both ends highlights variability in trip profitability and uneven revenue performance.

```python
#Chart 3: Profit Distribution Histogram

#Chart Type: Histogram
plt.figure(figsize=(8,6))
sns.histplot(df['Profit'], bins=30,kde=True)
plt.axvline(0, color='red', linestyle='--', label='Break-even')
plt.title('Profit Distribution per Trip', fontsize=16)
plt.xlabel('Profit', fontsize=12)
plt.ylabel('Number of Trips', fontsize=12)
plt.show()
```

# Interpretation – Profit Distribution Histogram

Shows the distribution of profit across trips.

Most trips are profitable, while a few trips experience losses.

Highlights opportunities to improve low-profit or loss-making trips.

```python
#Chart 4: Total Profit by Route

#Chart Type: Barplot
plt.figure(figsize=(15,6))
sns.barplot(x='To_Station', y='Profit', hue='To_Station',data=df,palette='Set2')
plt.title('Total Profit by Route', fontsize=16)
plt.xlabel('Route', fontsize=12)
plt.ylabel('Total Profit', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

# Interpretation – Total Profit by Route

Displays total profit for each route.

Highlights the most and least profitable routes.

Helps identify routes that need pricing or operational improvements.

```python
#Chart 5: Profit vs Passengers per Trip

#Chart Type: Scatterplot
plt.figure(figsize=(10,6))
sns.scatterplot(x='Passengers',y='Profit',hue='Ticket_Type',data=df, palette='Set2')
plt.title('Profit vs Passengers per Trip', fontsize=16)
plt.xlabel('Number of Passengers', fontsize=12)
plt.ylabel('Profit', fontsize=12)
plt.legend(title='Ticket Type', fontsize=10)
plt.tight_layout()
plt.show()
```

# Interpretation – Passengers vs Profit Scatter Plot
Shows the relationship between passengers and profit.

Higher passenger trips usually make more profit, but not always.

Highlights trips that need optimization to improve revenue.

```python
#Chart 6: Correlation Heatmap of Key Trip Metrics

#Chart Type: Correlation Heatmap
plt.figure(figsize=(10,8))
corr = df[['Passengers', 'Profit', 'Fare', 'Cost_per_passenger']].corr()
sns.heatmap(corr,annot=True)
plt.title("Correlation Heatmap of Key Trip Metrics")
plt.show()
```

# Interpretation – Correlation Heatmap of Key Trip Metrics
Shows relationships between passengers, profit, fare, and cost per passenger.

Profit is positively correlated with passengers and fare, and negatively correlated with cost per passenger.

Highlights which factors most impact trip profitability and efficiency.

```python
#Chart 7: Distribution of Trip Distance

#Chart Type: Histogram
plt.figure(figsize=(8,6))
sns.histplot(df['Distance_km'], bins=20,kde=True)
plt.title('Distribution of Trip Distance')
plt.xlabel('Distance (km)')
plt.ylabel('Number of Trips')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# Interpretation – Distance_km Histogram

Most trips fall within shorter distance ranges, indicating frequent short-distance travel

Fewer trips occur at longer distances, showing declining usage as distance increases

This pattern suggests the metro is primarily used for daily, short-commute travel

```python
#Chart 8: Profit Contribution by Remarks

#Chart Type: Pie Chart
profit_by_remarks = df.groupby('Remarks')['Profit'].sum()

plt.figure(figsize=(8, 8))
plt.pie(profit_by_remarks,labels=profit_by_remarks.index,autopct='%1.1f%%',startangle=90)
plt.title('Profit Contribution by Remarks')
plt.show()
```

# Interpretation - Profit Pie Chart

The pie chart shows the percentage contribution of total profit by each remark category.

Each slice represents how much profit is generated during Festival and Normal periods.

```python
#Chart 9: fare per km vs Cost per km

#Chart Type: Scatter Plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Fare_per_km',y='Cost_per_km',hue='Remarks',data=df)
plt.title('Fare per km vs Cost per km')
plt.xlabel('Fare per km')
plt.ylabel('Cost per km')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# Interpretation - fare per km vs Cost per km

The chart compares fare and cost per passenger to understand pricing efficiency.

Normalized values are used to ensure fair comparison across varying trip distances.

```python
#chart10 - Month-wise Profit Line Chart
#Line chart

df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
df['Month'] = df['Date'].dt.to_period('M')
monthly_profit = df.groupby('Month')['Profit'].sum()

plt.figure(figsize=(8, 6))
plt.plot(
    monthly_profit.index.astype(str),
    monthly_profit.values,
    marker='v'
)
plt.title('Month-wise Profit Trend')
plt.xlabel('Month')
plt.ylabel('Total Profit')
plt.xticks(rotation=90)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

# Interpretation – Month-wise Profit Line Chart

The line chart shows total profit aggregated for each month

Each point represents the overall profitability of metro operations in that month

```python
#Chart 11: Profit Distribution by Remarks

#Chart Type: Box Plot
plt.figure(figsize=(8, 6))
sns.boxplot(y='Profit', x='Remarks', data=df)
plt.title('Profit Distribution by Remarks')
plt.xlabel('Profit')
plt.ylabel('Remarks')
plt.grid(True, axis='x', linestyle='--', alpha=0.5)
plt.show()
```

# Interpretation - Profit Box Plot by Remarks
Festival periods show a higher median profit compared to normal days.

Profit variability is greater during festivals, indicating fluctuating demand.

Presence of outliers suggests occasional exceptionally high or low profit trips.

```python
#Chart 12: Pair plot  by Remarks

#Chart Type: Pair Plot
sns.pairplot(
    df[['Distance_km', 'Fare', 'Cost_per_passenger', 'Passengers', 'Profit', 'Remarks']],
    hue='Remarks',
    diag_kind='kde',
    palette='viridis'
)

plt.suptitle('Metro Trip Metrics and Profit Distribution by Remarks', y=1.02)
plt.show()
```

# Interpretation – Pair Plot by Remarks

Clear separation across remarks indicates that travel conditions influence fare, cost, and profit behavior.

Positive relationships between distance, fare, and profit are visible, but the strength of these relationships varies by remarks category.

Overlapping clusters and scattered patterns suggest that higher fare or longer distance does not always guarantee higher profit, highlighting the impact of operational costs and passenger volume.
